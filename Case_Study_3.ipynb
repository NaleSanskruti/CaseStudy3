```python
# Case Study 3
# Hospital Patient Management System
class Node:
    def __init__(self, patient_id, name, priority):
        self.patient_id = patient_id
        self.name = name
        self.priority = priority
        self.next = None


class Hospital:
    def __init__(self):
        self.head = None

    # Add patient according to priority
    def add_patient(self, patient_id, name, priority):
        new_node = Node(patient_id, name, priority)

        if self.head is None or priority < self.head.priority:
            new_node.next = self.head
            self.head = new_node
        else:
            temp = self.head

            while (temp.next is not None and
                   temp.next.priority <= priority):
                temp = temp.next

            new_node.next = temp.next
            temp.next = new_node

        print("Patient added successfully!")

    # Display waiting patients
    def display_patients(self):
        if self.head is None:
            print("No patients waiting.")
            return

        temp = self.head
        print("\n--- Patient Waiting List ---")

        while temp is not None:
            print("Patient ID:", temp.patient_id)
            print("Name:", temp.name)
            print("Priority:", temp.priority)
            print("----------------------")
            temp = temp.next

    # Search patient by ID
    def search_patient(self, patient_id):
        temp = self.head

        while temp is not None:
            if temp.patient_id == patient_id:
                print("Patient Found!")
                print("Name:", temp.name)
                print("Priority:", temp.priority)
                return

            temp = temp.next

        print("Patient not found.")

    # Treat the highest-priority patient
    def treat_patient(self):
        if self.head is None:
            print("No patients waiting.")
            return

        temp = self.head
        print("Next patient:", temp.name)
        print("Patient ID:", temp.patient_id)
        print("Patient selected for treatment.")

        self.head = self.head.next

    # Delete patient by ID
    def delete_patient(self, patient_id):
        if self.head is None:
            print("No patients in the list.")
            return

        if self.head.patient_id == patient_id:
            self.head = self.head.next
            print("Patient removed successfully.")
            return

        temp = self.head

        while temp.next is not None:
            if temp.next.patient_id == patient_id:
                temp.next = temp.next.next
                print("Patient removed successfully.")
                return

            temp = temp.next

        print("Patient not found.")


# Main program
hospital = Hospital()

while True:
    print("\n===== SMART HOSPITAL MANAGEMENT =====")
    print("1. Add Patient")
    print("2. Display Patients")
    print("3. Search Patient")
    print("4. Treat Next Patient")
    print("5. Delete Patient")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        patient_id = int(input("Enter patient ID: "))
        name = input("Enter patient name: ")
        priority = int(input("Enter priority (1-4): "))

        if priority < 1 or priority > 4:
            print("Enter priority between 1 and 4.")
        else:
            hospital.add_patient(patient_id, name, priority)

    elif choice == "2":
        hospital.display_patients()

    elif choice == "3":
        patient_id = int(input("Enter patient ID to search: "))
        hospital.search_patient(patient_id)

    elif choice == "4":
        hospital.treat_patient()

    elif choice == "5":
        patient_id = int(input("Enter patient ID to delete: "))
        hospital.delete_patient(patient_id)

    elif choice == "6":
        print("Exiting Hospital Management System.")
        break

    else:
        print("Invalid choice. Try again.")
```
